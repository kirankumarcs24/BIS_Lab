```python
import random

try:
    NUM_PARTICLES = int(input("Enter number of particles (e.g., 10): "))
    MAX_ITERATIONS = int(input("Enter max optimization rounds (e.g., 20): "))

    print("\n--- Set Traffic Light Boundaries (Seconds) ---")
    MIN_GREEN = int(input("Enter minimum allowed green time (e.g., 10): "))
    MAX_GREEN = int(input("Enter maximum allowed green time (e.g., 60): "))

    print("\n--- Set the Secret Perfect Timing Target ---")
    perfect_ns = int(input(f"Enter target North-South green time ({MIN_GREEN}-{MAX_GREEN}): "))
    perfect_ew = int(input(f"Enter target East-West green time ({MIN_GREEN}-{MAX_GREEN}): "))
except ValueError:
    print("Invalid input. Please enter numbers only. Falling back to defaults.")
    NUM_PARTICLES = 10
    MAX_ITERATIONS = 20
    MIN_GREEN = 10
    MAX_GREEN = 60
    perfect_ns = 35
    perfect_ew = 25

W = 0.5
C1 = 1.5
C2 = 1.5

def evaluate_traffic_delay(green_times):
    ns_green, ew_green = green_times
    delay = abs(ns_green - perfect_ns) + abs(ew_green - perfect_ew)
    return delay

particles_position = []
particles_velocity = []
pbest_position = []
pbest_score = []

for i in range(NUM_PARTICLES):
    ns_start = random.randint(MIN_GREEN, MAX_GREEN)
    ew_start = random.randint(MIN_GREEN, MAX_GREEN)

    particles_position.append([ns_start, ew_start])
    particles_velocity.append([0.0, 0.0])

    pbest_position.append([ns_start, ew_start])
    pbest_score.append(evaluate_traffic_delay([ns_start, ew_start]))

gbest_score = min(pbest_score)
gbest_index = pbest_score.index(gbest_score)
gbest_position = list(pbest_position[gbest_index])

print("\n" + "=" * 60)
print(f"Initial Swarm Best: NS={gbest_position[0]}s, EW={gbest_position[1]}s (Delay Score: {gbest_score})")
print("=" * 60)

for round_num in range(1, MAX_ITERATIONS + 1):
    for i in range(NUM_PARTICLES):
        for d in range(2):
            r1 = random.random()
            r2 = random.random()

            memory_pull = C1 * r1 * (pbest_position[i][d] - particles_position[i][d])
            peer_pull = C2 * r2 * (gbest_position[d] - particles_position[i][d])

            particles_velocity[i][d] = (W * particles_velocity[i][d]) + memory_pull + peer_pull
            particles_position[i][d] = particles_position[i][d] + particles_velocity[i][d]

            if particles_position[i][d] < MIN_GREEN:
                particles_position[i][d] = MIN_GREEN
            if particles_position[i][d] > MAX_GREEN:
                particles_position[i][d] = MAX_GREEN

            particles_position[i][d] = int(round(particles_position[i][d]))

        current_score = evaluate_traffic_delay(particles_position[i])

        if current_score < pbest_score[i]:
            pbest_score[i] = current_score
            pbest_position[i] = list(particles_position[i])

            if current_score < gbest_score:
                gbest_score = current_score
                gbest_position = list(particles_position[i])

    print(f"Round {round_num}: Best Timings Found -> NS={gbest_position[0]}s, EW={gbest_position[1]}s (Delay: {gbest_score})")

print("\n" + "=" * 60)
print("🎯 SEARCH COMPLETED")
print(f"Target Secret Timing:                 NS={perfect_ns}s, EW={perfect_ew}s")
print(f"Swarm's Final Optimized Configuration: NS={gbest_position[0]}s, EW={gbest_position[1]}s")
print(f"Final Traffic Jam Score (0 is ideal):  {gbest_score}")
print("=" * 60)
```
