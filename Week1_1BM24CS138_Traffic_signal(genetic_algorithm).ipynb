```python
import numpy as np

num_phases = 4
population_size = 50
generations = 30
mutation_rate = 0.1
min_green = 10
max_green = 60

traffic_demand = []
for i in range(num_phases):
    demand = float(input(f"Enter vehicle demand per minute for Phase {i+1}: "))
    traffic_demand.append(demand)
traffic_demand = np.array(traffic_demand)

def fitness(individual):
    total_green = np.sum(individual)
    cycle_time = total_green + len(individual) * 4
    delay = 0
    for i in range(num_phases):
        red_time = cycle_time - individual[i]
        vehicles_arrived = traffic_demand[i] * (cycle_time / 60.0)
        delay += vehicles_arrived * red_time
    return 1.0 / (1.0 + delay)

population = np.random.randint(min_green, max_green + 1, size=(population_size, num_phases))

for _ in range(generations):
    fit_scores = np.array([fitness(ind) for ind in population])
    probs = fit_scores / np.sum(fit_scores)

    selected_indices = np.random.choice(population_size, size=population_size, p=probs)
    selected = population[selected_indices]

    next_population = []
    for i in range(0, population_size, 2):
        p1 = selected[i]
        p2 = selected[(i + 1) % population_size]
        if np.random.rand() < 0.8:
            point = np.random.randint(1, num_phases)
            c1 = np.concatenate([p1[:point], p2[point:]])
            c2 = np.concatenate([p2[:point], p1[point:]])
        else:
            c1, c2 = p1.copy(), p2.copy()

        for c in [c1, c2]:
            for j in range(num_phases):
                if np.random.rand() < mutation_rate:
                    c[j] = np.random.randint(min_green, max_green + 1)
        next_population.append(c1)
        next_population.append(c2)

    population = np.array(next_population[:population_size])

fit_scores = np.array([fitness(ind) for ind in population])
best_index = np.argmax(fit_scores)
best_solution = population[best_index]
print("Optimal Green Times (seconds):", best_solution)
```
